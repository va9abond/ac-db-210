Задание 1.

Напишите SQL-запрос, возвращающий фамилии всех студентов с именем Александр или с баллом больше 4.5

```python
%%sql

SELECT lastname
FROM Students
WHERE name = "Александр" OR gpa >= 4.5
```

Задание 2.

Вывести только тех студентов, фамилия которых начинается с буквы 'к', а в имени есть буква 'е'.

```python
%%sql

SELECT name, lastname
FROM Students
WHERE lastname LIKE 'К%' AND name LIKE '%е%'
```

Задание 3.

Для каждой группы выведите оценку её лучшего студента. Результат отсортировать по убыванию баллов.

```python
%%sql

SELECT
    class,
    COUNT(id) as ids,
    MAX(gpa) as max_gpa,
    lastname || ' ' || name as max_gpa_student
FROM Students
WHERE class IS NOT NULL
GROUP BY class
ORDER BY max_gpa DESC;
```

Задание 4.

Вывести специальности (+ их средние баллы), количество групп в которых больше одной

```python
%%sql

select spec, avg(gpa) as avg_gpa
from Students
where spec IS NOT NULL
group by spec
```

```python
%%sql

SELECT class, COUNT(DISTINCT spec) as specs_count
FROM Students
WHERE
    spec IS NOT NULL AND
    class IS NOT NULL
GROUP BY class
```

Задание 5.

Для всех специальностей вывести средний балл по специальности среди только тех студентов, в чьей фамилии вторая или третья буква 'т', при условии, что количество таких студентов на специальности не менее 2. В обратном случае, специальность выводить не требуется.

```python
%%sql

SELECT spec, AVG(gpa) as avg_gpa
FROM Students
WHERE
    spec IS NOT NULL AND
    (lastname LIKE '_т%' OR lastname LIKE '__т%')
GROUP BY spec
```

```python
%%sql

SELECT spec, lastname, gpa
FROM Students
WHERE
    spec IS NOT NULL AND
    (lastname LIKE '_т%' OR lastname LIKE '__т%')
ORDER BY spec ASC
```

```python
%%sql

SELECT spec, AVG(gpa) as avg_gpa
FROM Students
WHERE
    spec IS NOT NULL AND
    (lastname LIKE '_т%' OR lastname LIKE '__т%')
GROUP BY spec
HAVING COUNT(id) >= 2
```

```python
%%sql

SELECT *
FROM (
    SELECT spec, lastname, gpa
    FROM Students
    WHERE 
        spec IS NOT NULL AND
        (lastname LIKE '_т%' OR lastname LIKE '__т%')
    ORDER BY spec ASC
) as temp
```

```python
%%sql

SELECT *
FROM (
    SELECT spec, SUM(gpa) as sum_gpa, COUNT(id) as count_id
    FROM Students
    WHERE 
        spec IS NOT NULL AND
        (lastname LIKE '_т%' OR lastname LIKE '__т%')
    GROUP BY spec
) as temp
```

```python
%%sql

SELECT temp.spec, temp.sum_gpa / temp.count_id as avg_gpa
FROM (
    SELECT spec, SUM(gpa) as sum_gpa, COUNT(id) as count_id
    FROM Students
    WHERE 
        spec IS NOT NULL AND
        (lastname LIKE '_т%' OR lastname LIKE '__т%')
    GROUP BY spec
) as temp
WHERE temp.count_id >= 2
```
